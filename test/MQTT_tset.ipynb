```python
import base64, numpy as np
from paho.mqtt import client as mqtt

BROKER = "192.168.100.20"
PORT   = 1883
TOPIC  = "fl/+/weight"

weights = {}  # 디바이스별 최신 가중치 저장

def on_connect(client, userdata, flags, rc):
    print("connected:", rc)
    client.subscribe(TOPIC)

def on_message(client, userdata, msg):
    dev = msg.topic.split("/")[1] if msg.topic.count("/") >= 2 else "unknown"
    raw = base64.b64decode(msg.payload)         # bytes → raw
    arr = np.frombuffer(raw, dtype="<f4")       # float32 (LE)
    weights[dev] = arr
    print(f"{dev}: {arr.size} floats, \n first5: {arr[:5]}, \n last5: {arr[-5:]}, \n sum total weights: {np.sum(arr)}")

client = mqtt.Client()
client.on_connect = on_connect
client.on_message = on_message
client.connect(BROKER, PORT, keepalive=60)
client.loop_forever()   # 여기서 계속 대기하며 수신
```
