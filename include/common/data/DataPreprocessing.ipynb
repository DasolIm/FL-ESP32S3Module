```python
from sklearn.datasets import fetch_openml

# 'letter' 데이터셋 이름과 version=1을 지정
X, y = fetch_openml(name='letter', version=1, return_X_y=True)
```

```python
import pandas as pd

data = pd.DataFrame(X)
data['class'] = y

data
```

```python
data.to_csv('./letter_raw.csv', index=False)
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(data, test_size=0.2, random_state=42)
train
```

```python
import numpy as np

data_ls = np.array_split(train, 5)
data_ls[0]
```

```python
for i, data in enumerate(data_ls):
    data.to_csv(f'./letter_train_{i}.csv', index=False)
    print(f"Subset {i}: {data.shape[0]}, {data.shape[1]}")
```

```python
import pandas as pd

def csv_to_header(csv_path: str, header_path: str):
    """
    CSV → C header (.h) with 1D PROGMEM arrays and aitensor_t definitions.
    x_data: SHAPE0*SHAPE1 flattened, y_labels: SHAPE0*SHAPE2 flattened one-hot.
    """
    df = pd.read_csv(csv_path)
    n_samples, n_cols = df.shape
    n_features = n_cols - 1    # 마지막 컬럼(레이블) 제외
    n_classes  = 26

    # flatten feature data
    flat_x = df.iloc[:, :n_features].to_numpy().astype(int).flatten()
    # build one-hot flattened labels
    flat_y = []
    for label in df.iloc[:, n_features]:
        idx = ord(str(label).upper()) - ord('A')
        for i in range(n_classes):
            flat_y.append(1.0 if i == idx else 0.0)

    with open(header_path, 'w') as f:
        # Shape 정의
        f.write(f'#define N_DATAS_VALUE {n_samples}\n')
        f.write(f'#define INPUT_SHAPE {n_features}\n')
        f.write(f'#define TARGET_SHAPE {n_classes}\n\n')

        # x_data 1D 배열
        f.write(f'const float x_data[N_DATAS_VALUE * INPUT_SHAPE] PROGMEM = {{\n')
        for i, v in enumerate(flat_x):
            if i % n_features == 0:
                f.write('    ')
            f.write(f'{v/16}, ')
            if (i + 1) % n_features == 0:
                f.write('\n')
        f.write('};\n\n')

        # y_labels 1D one-hot 배열
        f.write(f'const float y_labels[N_DATAS_VALUE * TARGET_SHAPE] PROGMEM = {{\n')
        for i, v in enumerate(flat_y):
            if i % n_classes == 0:
                f.write('    ')
            f.write(f'{v:.1f}f, ')
            if (i + 1) % n_classes == 0:
                f.write('\n')
        f.write('};\n\n')

#         f.write('''
# const size_t N_DATAS PROGMEM = N_DATAS_VALUE;
# // shape 배열 정의
# const size_t x_shape[2] PROGMEM = { N_DATAS, INPUT_SHAPE };
# const size_t y_shape[2] PROGMEM = { N_DATAS, TARGET_SHAPE };
#
# // aitensor_t 객체 정의
# aitensor_t x_tensor PROGMEM = AITENSOR_2D_F32(x_shape, x_data);
# aitensor_t y_tensor PROGMEM = AITENSOR_2D_F32(y_shape, y_labels);
#         ''')
```

```python
for i, _ in enumerate(data_ls):
    csv_to_header(f'./letter_train_{i}.csv', f'./data_{i}.h')
    print(f"Subset {i} header generated.")
```
